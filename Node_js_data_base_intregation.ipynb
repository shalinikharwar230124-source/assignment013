**3 Define a sequelize model for a Task that has a title(string) and a completed (boolean ) field**
-

```python
const { Sequelize, DataTypes } = require('sequelize');

// Create Sequelize instance
const sequelize = new Sequelize('app_db', 'root', 'password', {
  host: 'localhost',
  dialect: 'mysql'
});

// Define Task model
const Task = sequelize.define('Task', {
  title: {
    type: DataTypes.STRING,
    allowNull: false
  },
  completed: {
    type: DataTypes.BOOLEAN,
    defaultValue: false
  }
});

// Export model (optional)
module.exports = Task;
```

### Notes:
- Sequelize is an ORM used for interacting with databases in Node.js.
- DataTypes.STRING is used for text fields.
- DataTypes.BOOLEAN is used for true/false values.
- defaultValue sets the initial value of the completed field.

**6 write the code to connect to a local MongoDB server and a database named innentroy using the offical mongodb driver**
--MongoDB is a NoSQL database, and the official MongoDB Node.js driver is used to connect and interact with it. To establish a connection, we use the MongoClient class from the mongodb package.

The following code connects to a local MongoDB server and accesses a database named "inventory".

```python
// Connection URL
const url = 'mongodb://127.0.0.1:27017';

// Database Name
const dbName = 'inventory';

// Create MongoDB client
const client = new MongoClient(url);

async function connectDB() {
  try {
    // Connect to server
    await client.connect();
    console.log('Connected to MongoDB server');

    // Connect to database
    const db = client.db(dbName);
    console.log(Connected to database: ${dbName});

  } catch (error) {
    console.error('Connection failed:', error.message);
  } finally {
    await client.close();
  }
}

// Call function
connectDB();
```

### Notes:
- MongoClient is used to connect to the MongoDB server.
- The URL 'mongodb://127.0.0.1:27017' represents the local MongoDB server.
- The db() method is used to access a specific database.
- async/await is used to handle asynchronous operations.
- The connection is closed using client.close() after use.

**7 Using the native MongoDB driver, insert a new document into a users collection with a name of "John Doe" and an age of 30.

### Answer:
MongoDB is a NoSQL database that stores data in the form of documents. The official MongoDB Node.js driver provides methods to interact with the database, including inserting documents into collections.

To insert a new document into a collection, the insertOne() method is used. The following code connects to the MongoDB database and inserts a document into the "users" collection.

```python
const { MongoClient } = require('mongodb');

// Connection URL
const url = 'mongodb://127.0.0.1:27017';

// Database Name
const dbName = 'inventory';

const client = new MongoClient(url);

async function insertUser() {
  try {
    // Connect to MongoDB
    await client.connect();
    console.log('Connected to MongoDB');

    // Select database
    const db = client.db(dbName);

    // Select collection
    const collection = db.collection('users');

    // Insert document
    const result = await collection.insertOne({
      name: "John Doe",
      age: 30
    });

    console.log('User inserted with ID:', result.insertedId);

  } catch (error) {
    console.error('Error inserting document:', error.message);
  } finally {
    await client.close();
  }
}

// Call function
insertUser();
```

### Notes:
- MongoClient is used to connect to the MongoDB server.
- insertOne() method inserts a single document into the collection.
- The document contains fields: name and age.
- async/await is used to handle asynchronous database operations.
- The connection is closed using client.close() after execution.

**8 Write a MongoDB query to find all users in a collection who are older than 25.

### Answer:
MongoDB provides powerful query capabilities to retrieve documents based on specific conditions. To find documents that satisfy a condition, the find() method is used along with query operators.

In this case, we use the $gt (greater than) operator to fetch all users whose age is greater than 25

```python
const { MongoClient } = require('mongodb');

const url = 'mongodb://127.0.0.1:27017';
const client = new MongoClient(url);

async function findUsers() {
  try {
    // Connect to MongoDB
    await client.connect();
    console.log('Connected to MongoDB');

    // Select database
    const db = client.db('inventory');

    // Query users older than 25
    const users = await db.collection('users')
      .find({ age: { $gt: 25 } })
      .toArray();

    console.log('Users older than 25:', users);

  } catch (error) {
    console.error('Error fetching users:', error.message);
  } finally {
    await client.close();
  }
}

// Call function
findUsers();
```

### Notes:
- find() method is used to retrieve documents from a collection.
- $gt is a MongoDB operator that means "greater than".
- The condition { age: { $gt: 25 } } filters users older than 25.
- toArray() converts the result into an array.
- async/await is used for handling asynchronous operations.

**9 Write a MongoDB query to update a user's age to 31, finding the user by their name, "John Doe".

### Answer:
MongoDB provides update operations to modify existing documents in a collection. The updateOne() method is used to update a single document based on a condition.

In this query, we find the user by name "John Doe" and update their age to 31 using the $set operator.

```python
const url = 'mongodb://127.0.0.1:27017';
const client = new MongoClient(url);

async function updateUser() {
  try {
    // Connect to MongoDB
    await client.connect();
    console.log('Connected to MongoDB');

    const db = client.db('inventory');

    // Update user age
    const result = await db.collection('users').updateOne(
      { name: "John Doe" },
      { $set: { age: 31 } }
    );

    console.log('User updated:', result.modifiedCount);

  } catch (error) {
    console.error('Error updating user:', error.message);
  } finally {
    await client.close();
  }
}

// Call function
updateUser();
```
